```python
import pandas as pd
import os

input_dir = "../data/03-11"
output_dir = "../data/parquets/03-11"
os.makedirs(output_dir, exist_ok=True)

csv_files = [
    "LDAP.csv",
    "MSSQL.csv",
    "NetBIOS.csv",
    "Portmap.csv",
    "Syn.csv",
    "UDP.csv",
    "UDPLag.csv",
]

chunk_size = 100_000  # Adjust this if you still get memory errors

for file in csv_files:
    file_path = os.path.join(input_dir, file)
    output_file = os.path.join(output_dir, file.replace(".csv", ".parquet"))

    print(f"Processing {file_path} ...")

    chunk_dfs = []  # temporarily store chunks before concatenating & saving

    for chunk in pd.read_csv(file_path, chunksize=chunk_size, low_memory=False):
        if "SimillarHTTP" in chunk.columns:
            chunk["SimillarHTTP"] = chunk["SimillarHTTP"].astype(str)
        chunk_dfs.append(chunk)

    # Merge all chunks (or write them one by one if needed)
    df = pd.concat(chunk_dfs, ignore_index=True)
    df.to_parquet(output_file, index=False, compression="snappy")
    del df, chunk_dfs  # free memory
    print(f"Saved to {output_file}")
```

```python
from scapy.all import PcapReader, IP, TCP, UDP, ICMP
import pandas as pd
from tqdm import tqdm
import math
from collections import Counter
import os
import pyarrow as pa
import pyarrow.parquet as pq


def calculate_entropy(data):
    """Calculate Shannon entropy of payload data"""
    if not data or len(data) == 0:
        return 0

    if isinstance(data, str):
        data = data.encode()

    byte_counts = Counter(data)
    data_len = len(data)

    entropy = 0
    for count in byte_counts.values():
        probability = count / data_len
        if probability > 0:
            entropy -= probability * math.log2(probability)

    return entropy


def extract_enhanced_packet_features(packet, flow_packet_counts, previous_timestamps):
    if IP not in packet:
        return None

    ip_layer = packet[IP]

    tcp_layer = packet.getlayer(TCP)
    udp_layer = packet.getlayer(UDP)
    icmp_layer = packet.getlayer(ICMP)

    src_port = None
    dst_port = None
    if tcp_layer:
        src_port = tcp_layer.sport
        dst_port = tcp_layer.dport
    elif udp_layer:
        src_port = udp_layer.sport
        dst_port = udp_layer.dport

    flow_id = f"{ip_layer.src}_{ip_layer.dst}_{src_port}_{dst_port}_{ip_layer.proto}"

    if flow_id not in flow_packet_counts:
        flow_packet_counts[flow_id] = 0
    flow_packet_counts[flow_id] += 1
    packet_position = flow_packet_counts[flow_id]

    current_time = packet.time
    inter_arrival_time = 0
    if flow_id in previous_timestamps:
        inter_arrival_time = current_time - previous_timestamps[flow_id]
    previous_timestamps[flow_id] = current_time

    payload = None
    if tcp_layer and tcp_layer.payload:
        payload = bytes(tcp_layer.payload)
    elif udp_layer and udp_layer.payload:
        payload = bytes(udp_layer.payload)
    elif icmp_layer and icmp_layer.payload:
        payload = bytes(icmp_layer.payload)

    payload_entropy = calculate_entropy(payload) if payload else 0

    packet_direction = (
        "forward"
        if (src_port and src_port < 1024) or (dst_port and dst_port >= 1024)
        else "backward"
    )

    features = {
        # Basic packet features
        "timestamp": packet.time,
        "src_ip": ip_layer.src,
        "dst_ip": ip_layer.dst,
        "src_port": src_port,
        "dst_port": dst_port,
        "protocol": ip_layer.proto,
        "length": len(packet),
        "ttl": ip_layer.ttl,
        "payload_size": len(payload) if payload else 0,
        # IP-level features
        "ip_id": ip_layer.id,
        "ip_flags": int(ip_layer.flags),
        "ip_fragment_offset": ip_layer.frag,
        "ip_tos": ip_layer.tos,
        # TCP-specific features
        "tcp_flags": int(tcp_layer.flags) if tcp_layer else None,
        "tcp_window": tcp_layer.window if tcp_layer else None,
        "tcp_seq": tcp_layer.seq if tcp_layer else None,
        "tcp_ack": tcp_layer.ack if tcp_layer else None,
        "tcp_urgent": tcp_layer.urgptr if tcp_layer else None,
        # ICMP-specific features
        "icmp_type": icmp_layer.type if icmp_layer else None,
        "icmp_code": icmp_layer.code if icmp_layer else None,
        # Flow and temporal features
        "flow_id": flow_id,
        "packet_position": packet_position,
        "inter_arrival_time": inter_arrival_time,
        # Payload characteristics
        "payload_entropy": payload_entropy,
        "packet_direction": packet_direction,
    }

    return features


def get_consistent_schema():
    """Define a consistent schema for all parquet files"""
    return pa.schema(
        [
            ("timestamp", pa.float64()),
            ("src_ip", pa.string()),
            ("dst_ip", pa.string()),
            ("src_port", pa.float64()),  # float64 to handle None values
            ("dst_port", pa.float64()),
            ("protocol", pa.int64()),
            ("length", pa.int64()),
            ("ttl", pa.int64()),
            ("payload_size", pa.int64()),
            ("ip_id", pa.int64()),
            ("ip_flags", pa.int64()),
            ("ip_fragment_offset", pa.int64()),
            ("ip_tos", pa.int64()),
            ("tcp_flags", pa.float64()),  # float64 to handle None values
            ("tcp_window", pa.float64()),
            ("tcp_seq", pa.float64()),
            ("tcp_ack", pa.float64()),
            ("tcp_urgent", pa.float64()),
            ("icmp_type", pa.float64()),
            ("icmp_code", pa.float64()),
            ("flow_id", pa.string()),
            ("packet_position", pa.int64()),
            ("inter_arrival_time", pa.float64()),
            ("payload_entropy", pa.float64()),
            ("packet_direction", pa.string()),
        ]
    )


def process_pcap_to_parquet(pcap_path, output_path, chunk_size=30000):
    """
    Process a PCAP file and save extracted features to a single parquet file.

    Args:
        pcap_path (str): Path to the input PCAP file
        output_path (str): Directory path where the parquet file will be saved
        chunk_size (int): Number of packets to process before saving to parquet
    """

    # Get PCAP filename without extension for parquet filename
    pcap_filename = os.path.basename(pcap_path)
    pcap_name = os.path.splitext(pcap_filename)[0]
    parquet_file = os.path.join(output_path, f"{pcap_name}.parquet")

    rows = []
    packet_count = 0
    flow_packet_counts = {}
    previous_timestamps = {}
    parquet_writer = None
    schema = get_consistent_schema()

    print(f"Processing PCAP: {pcap_path}")
    print(f"Output parquet: {parquet_file}")

    with PcapReader(pcap_path) as pcap:
        for pkt in tqdm(pcap, desc="Reading PCAP"):
            info = extract_enhanced_packet_features(
                pkt, flow_packet_counts, previous_timestamps
            )
            if info:
                rows.append(info)
                packet_count += 1

            if len(rows) >= chunk_size:
                df = pd.DataFrame(rows)

                # Ensure consistent data types
                df = df.astype(
                    {
                        "timestamp": "float64",
                        "src_ip": "string",
                        "dst_ip": "string",
                        "src_port": "float64",
                        "dst_port": "float64",
                        "protocol": "int64",
                        "length": "int64",
                        "ttl": "int64",
                        "payload_size": "int64",
                        "ip_id": "int64",
                        "ip_flags": "int64",
                        "ip_fragment_offset": "int64",
                        "ip_tos": "int64",
                        "tcp_flags": "float64",
                        "tcp_window": "float64",
                        "tcp_seq": "float64",
                        "tcp_ack": "float64",
                        "tcp_urgent": "float64",
                        "icmp_type": "float64",
                        "icmp_code": "float64",
                        "flow_id": "string",
                        "packet_position": "int64",
                        "inter_arrival_time": "float64",
                        "payload_entropy": "float64",
                        "packet_direction": "string",
                    }
                )

                table = pa.Table.from_pandas(df, schema=schema)

                # Initialize writer on first chunk
                if parquet_writer is None:
                    parquet_writer = pq.ParquetWriter(parquet_file, schema)

                # Write chunk to parquet (memory efficient)
                parquet_writer.write_table(table)

                print(f"Saved chunk with {len(rows)} packets (Total: {packet_count})")
                rows = []

    # Save remaining packets
    if rows:
        df = pd.DataFrame(rows)

        # Ensure consistent data types
        df = df.astype(
            {
                "timestamp": "float64",
                "src_ip": "string",
                "dst_ip": "string",
                "src_port": "float64",
                "dst_port": "float64",
                "protocol": "int64",
                "length": "int64",
                "ttl": "int64",
                "payload_size": "int64",
                "ip_id": "int64",
                "ip_flags": "int64",
                "ip_fragment_offset": "int64",
                "ip_tos": "int64",
                "tcp_flags": "float64",
                "tcp_window": "float64",
                "tcp_seq": "float64",
                "tcp_ack": "float64",
                "tcp_urgent": "float64",
                "icmp_type": "float64",
                "icmp_code": "float64",
                "flow_id": "string",
                "packet_position": "int64",
                "inter_arrival_time": "float64",
                "payload_entropy": "float64",
                "packet_direction": "string",
            }
        )

        table = pa.Table.from_pandas(df, schema=schema)

        if parquet_writer is None:
            parquet_writer = pq.ParquetWriter(parquet_file, schema)

        parquet_writer.write_table(table)
        print(f"Saved final chunk with {len(rows)} packets")

    # Close the writer
    if parquet_writer:
        parquet_writer.close()

    print(f"Done. Processed {packet_count} packets.")
    print(f"Total flows encountered: {len(flow_packet_counts)}")
    print(f"Output saved to: {parquet_file}")

    # # Print sample of extracted features
    # if packet_count > 0:
    #     print("\nSample of extracted features:")
    #     sample_df = pd.read_parquet(parquet_file)  # Only read first 5 rows
    #     print(sample_df.head())
    #     print(f"\nColumns: {list(sample_df.columns)}")

    #     # Get total row count without loading all data
    #     parquet_info = pq.read_metadata(parquet_file)
    #     total_rows = parquet_info.num_rows
    #     print(f"Total rows in parquet: {total_rows}")


def process_multiple_pcap_files(input_dir, output_dir, chunk_size=30000):
    """
    Process all PCAP files in a directory and save each as a separate parquet file.

    Args:
        input_dir (str): Directory containing PCAP files
        output_dir (str): Directory where parquet files will be saved
        chunk_size (int): Number of packets to process before saving to parquet
    """

    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)

    # Get all files in the input directory
    pcap_files = [
        f for f in os.listdir(input_dir) if os.path.isfile(os.path.join(input_dir, f))
    ]

    if not pcap_files:
        print(f"No files found in {input_dir}")
        return

    print(f"Found {len(pcap_files)} files to process")
    print(f"Input directory: {input_dir}")
    print(f"Output directory: {output_dir}")
    print("-" * 50)

    successful_files = 0
    failed_files = []

    for i, filename in enumerate(pcap_files, 1):
        pcap_path = os.path.join(input_dir, filename)

        print(f"\n[{i}/{len(pcap_files)}] Processing: {filename}")

        try:
            process_pcap_to_parquet(pcap_path, output_dir, chunk_size)
            successful_files += 1
            print(f"✓ Successfully processed: {filename}")

        except Exception as e:
            print(f"✗ Failed to process {filename}: {str(e)}")
            failed_files.append(filename)
            continue

    print("\n" + "=" * 50)
    print("PROCESSING SUMMARY")
    print("=" * 50)
    print(f"Total files: {len(pcap_files)}")
    print(f"Successfully processed: {successful_files}")
    print(f"Failed: {len(failed_files)}")

    if failed_files:
        print(f"\nFailed files:")
        for failed_file in failed_files:
            print(f"  - {failed_file}")
```

```python
import os
import pandas as pd


def count_none_flow_ids(parquet_dir):
    total_none_count = 0
    print(f"Checking for 'None' in Flow ID across parquet files in: {parquet_dir}\n")

    for file in os.listdir(parquet_dir):
        if file.endswith(".parquet"):
            file_path = os.path.join(parquet_dir, file)
            try:
                df = pd.read_parquet(file_path, columns=["Flow ID"])
                none_count = df["Flow ID"].str.contains("None", na=False).sum()
                print(f"{file}: {none_count} Flow IDs containing 'None'")
                total_none_count += none_count
            except Exception as e:
                print(f"Error reading {file}: {e}")

    print(f"\nTotal 'None' Flow IDs across all files: {total_none_count}")

count_none_flow_ids("/home/AGFirass/Documents/Github/Transformer-Based-DDoS-Detection/data/flow_level/03-11")
```

```python
import os
import pandas as pd
import pyarrow.parquet as pq

def reformat_flow_ids(input_dir, output_dir=None):
    if output_dir is None:
        output_dir = input_dir
    else:
        os.makedirs(output_dir, exist_ok=True)

    print(f"\nReformatting Flow IDs in folder: {input_dir}")
    print(f"Output directory: {output_dir}")
    print("-" * 60)

    for filename in os.listdir(input_dir):
        if filename.endswith(".parquet"):
            input_path = os.path.join(input_dir, filename)
            df = pd.read_parquet(input_path)

            if "Flow ID" not in df.columns:
                print(f"✗ Skipping {filename}: 'Flow ID' column not found.")
                continue

            df["Flow ID"] = df["Flow ID"].str.replace("-", "_", regex=False)

            output_path = os.path.join(output_dir, filename)
            df.to_parquet(output_path, index=False)
            print(f"✓ Reformatted and saved: {filename}")

    print("\nAll Flow IDs have been reformatted.")

reformat_flow_ids(
    input_dir="/home/AGFirass/Documents/Github/Transformer-Based-DDoS-Detection/data/flow_level/03-11"
)
```

Filtering and downsampling the flow-level data

```python
def filter_and_sample_flow_files(input_dir, output_dir, attack_sample_limit=2500):
    os.makedirs(output_dir, exist_ok=True)

    files = [f for f in os.listdir(input_dir) if f.endswith(".parquet")]

    for filename in files:
        file_path = os.path.join(input_dir, filename)
        print(f"Processing {filename} ...")

        df = pd.read_parquet(file_path)

        # Trim spaces in column names
        df.columns = df.columns.str.strip()

        attack_label = os.path.splitext(filename)[0].lower()

        # Filter for benign and attack label rows
        benign_rows = df[df["Label"].str.lower() == "benign"]
        attack_rows = df[df["Label"].str.lower() == attack_label]

        # Sample attack rows if more than limit
        if len(attack_rows) > attack_sample_limit:
            attack_rows = attack_rows.sample(n=attack_sample_limit, random_state=42)

        # Concatenate benign + sampled attack
        filtered_df = pd.concat([benign_rows, attack_rows], ignore_index=True)

        print(f" - Original rows: {len(df)}")
        print(f" - BENIGN rows kept: {len(benign_rows)}")
        print(f" - {attack_label} rows kept (sampled if needed): {len(attack_rows)}")
        print(f" - Total rows after filtering and sampling: {len(filtered_df)}")

        output_path = os.path.join(output_dir, filename)
        filtered_df.to_parquet(output_path, index=False)

        print(f" --> Saved filtered data to {output_path}\n")


input_dir = "/home/AGFirass/Documents/Github/Transformer-Based-DDoS-Detection/data/flow_level/03-11"
output_dir = "/home/AGFirass/Documents/Github/Transformer-Based-DDoS-Detection/data/flow_level/filtered_03-11"
filter_and_sample_flow_files(input_dir, output_dir, attack_sample_limit=2500)
```

```python
import os
import pandas as pd


def load_all_flow_ids(flow_level_dir, flow_files, save_csv_path):
    flow_ids = set()
    for f in flow_files:
        path = os.path.join(flow_level_dir, f)
        df = pd.read_parquet(path)

        df.columns = df.columns.str.strip().str.lower()

        flow_id_col = next(
            (col for col in df.columns if col.replace(" ", "_") == "flow_id"), None
        )
        if flow_id_col is None:
            raise ValueError(f"'Flow ID' column not found in {f}")

        flow_ids.update(df[flow_id_col].unique())

    # Save to CSV
    flow_ids_list = list(flow_ids)
    flow_ids_df = pd.DataFrame(flow_ids_list, columns=["flow_id"])
    flow_ids_df.to_csv(save_csv_path, index=False)
    print(f"Saved {len(flow_ids)} unique flow_ids to {save_csv_path}")

    return flow_ids


def filter_packet_files(packet_dir, packet_files, flow_ids, output_dir=None):
    if output_dir:
        os.makedirs(output_dir, exist_ok=True)

    flow_ids_set = set(flow_ids)

    for f in packet_files:
        path = os.path.join(packet_dir, f)
        print(f"Processing {f} ...")

        df = pd.read_parquet(path)

        df.columns = df.columns.str.strip().str.lower()

        if "flow_id" not in df.columns:
            print(f"Warning: 'flow_id' column not found in {f}, skipping file.")
            continue

        filtered_df = df[df["flow_id"].isin(flow_ids_set)]

        print(f" - Original packets: {len(df)}, Filtered packets: {len(filtered_df)}")

        if output_dir:
            out_path = os.path.join(output_dir, f)
            filtered_df.to_parquet(out_path, index=False)
            print(f" --> Saved filtered packets to {out_path}")
        else:
            # Overwrite original file (be cautious)
            filtered_df.to_parquet(path, index=False)
            print(f" --> Overwrote original file with filtered packets")
```

Verify the total of packets for each flow VS the total from the packet level data

```python
def compare_filtered_flow_vs_packet_packets(
    flow_dir="/home/AGFirass/Documents/Github/Transformer-Based-DDoS-Detection/data/flow_level/filtered_03-11",
    packet_dir="/home/AGFirass/Documents/Github/Transformer-Based-DDoS-Detection/data/packet_level/filtered_03-11",
    packet_files=None,
):
    if packet_files is None:
        packet_files = generate_packet_filenames()

    flow_files = [f for f in os.listdir(flow_dir) if f.endswith(".parquet")]

    flow_packet_counts = {}

    for flow_file in flow_files:
        flow_path = os.path.join(flow_dir, flow_file)
        flow_df = pd.read_parquet(flow_path)
        flow_df.columns = flow_df.columns.str.strip()

        flow_df["total_packets_flow"] = (
            flow_df["Total Fwd Packets"] + flow_df["Total Backward Packets"]
        )

        for flow_id, pkt_count in zip(
            flow_df["Flow ID"], flow_df["total_packets_flow"]
        ):
            flow_packet_counts[flow_id] = flow_packet_counts.get(flow_id, 0) + pkt_count

    packet_counts = {}

    for pf in packet_files:
        pf_path = os.path.join(packet_dir, pf)
        if not os.path.exists(pf_path):
            print(f"Warning: Packet file {pf_path} does not exist, skipping.")
            continue
        packet_df = pd.read_parquet(pf_path, columns=["flow_id"])
        packet_df.columns = packet_df.columns.str.strip()

        counts = packet_df["flow_id"].value_counts()
        for flow_id, count in counts.items():
            packet_counts[flow_id] = packet_counts.get(flow_id, 0) + count

    differences = []
    for flow_id, flow_pkt_count in flow_packet_counts.items():
        pkt_count = packet_counts.get(flow_id, 0)
        diff = pkt_count - flow_pkt_count
        differences.append(
            {
                "Flow ID": flow_id,
                "Flow Packet Count": flow_pkt_count,
                "Packet Packet Count": pkt_count,
                "Difference": diff,
            }
        )

    diff_df = pd.DataFrame(differences)

    print(f"Total flows compared: {len(diff_df)}")
    print(f"Perfect matches (Difference=0): {(diff_df['Difference'] == 0).sum()}")
    print(
        f"Positive differences (packet count > flow count): {(diff_df['Difference'] > 0).sum()}"
    )
    print(
        f"Negative differences (packet count < flow count): {(diff_df['Difference'] < 0).sum()}"
    )

    return diff_df


diff_df = compare_filtered_flow_vs_packet_packets()

diff_df.to_csv(
    "/home/AGFirass/Documents/Github/Transformer-Based-DDoS-Detection/data/flow_vs_packet_differences.csv",
    index=False,
)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from pandas.plotting import radviz
import seaborn as sns

plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
```

```python
# Configuration and data paths
data_folder = "/home/AGFirass/Documents/Github/Transformer-Based-DDoS-Detection/data/flow_level/filtered_03-11"

drop_cols = [
    "Unnamed: 0",
    "Flow ID",
    "Source IP",
    "Destination IP",
    "Timestamp",
    "SimilarityHTTP",
    "Source Port",
    "Destination Port",
]

# Create output directory for plots
os.makedirs("radviz_plots", exist_ok=True)
```

```python
# Cell 3: Load benign data
benign_path = os.path.join(data_folder, "BENIGN.parquet")
df_benign = pd.read_parquet(benign_path)
df_benign["Label"] = 0

print(f"Loaded benign data: {len(df_benign)} samples")
print(f"Benign data shape: {df_benign.shape}")
```

```python
# Get list of attack files
all_files = [
    f
    for f in os.listdir(data_folder)
    if f.endswith(".parquet") and f != "BENIGN.parquet"
]

print(f"Found {len(all_files)} attack files:")
for file in all_files:
    print(f"  - {file}")
```

```python
def clean_data(df, drop_cols):
    df_clean = df.copy()
    for col in drop_cols:
        if col in df_clean.columns:
            df_clean = df_clean.drop(columns=[col])
    return df_clean

def prepare_features(X):
    # Convert categorical (Protocol)
    if "Protocol" in X.columns:
        X["Protocol"] = X["Protocol"].astype(str)
        X = pd.get_dummies(X, columns=["Protocol"])
    return X

def prepare_radviz_data(X, y, top_features, attack_name):
    # Select top features for RadViz
    X_radviz = X[top_features].copy()

    # Handle infinite values and NaN
    X_radviz = X_radviz.replace([np.inf, -np.inf], np.nan)
    X_radviz = X_radviz.fillna(X_radviz.median())

    # Scale features for better visualization
    scaler = StandardScaler()
    X_scaled = pd.DataFrame(scaler.fit_transform(X_radviz), columns=X_radviz.columns)

    # Add labels for RadViz
    X_scaled["Class"] = y.map({0: "Benign", 1: attack_name})

    return X_scaled
```

```python
# Main processing loop - Load and process each attack
results = {}

for attack_file in all_files:
    attack_name = attack_file.replace(".parquet", "")
    print(f"\n{'='*60}")
    print(f"Processing attack: {attack_name}")
    print(f"{'='*60}")

    # Load attack data, add label 1
    attack_path = os.path.join(data_folder, attack_file)
    df_attack = pd.read_parquet(attack_path)
    df_attack["Label"] = 1

    print(f"Loaded attack data: {len(df_attack)} samples")

    # Create copies to avoid modifying original data
    df_benign_copy = clean_data(df_benign, drop_cols)
    df_attack_copy = clean_data(df_attack, drop_cols)

    # Combine benign + attack data
    df_combined = pd.concat([df_benign_copy, df_attack_copy], ignore_index=True)
    print(f"Combined dataset: {len(df_combined)} samples")

    # Separate features and target
    X = df_combined.drop(columns=["Label"])
    y = df_combined["Label"]

    # Convert categorical columns if any
    X = prepare_features(X)

    # Store results for this attack
    results[attack_name] = {
        "X": X,
        "y": y,
        "benign_samples": len(df_benign_copy),
        "attack_samples": len(df_attack_copy),
        "total_samples": len(df_combined),
        "features": X.columns.tolist(),
    }

    print(f"Features after preprocessing: {X.shape[1]}")
```

```python
for attack_name in results.keys():
    print(f"\nTraining Random Forest for {attack_name}...")

    X = results[attack_name]["X"]
    y = results[attack_name]["y"]

    # Drop non-numeric columns
    non_numeric_cols = X.select_dtypes(include=["object"]).columns.tolist()
    if non_numeric_cols:
        print(f"Dropping non-numeric columns for {attack_name}: {non_numeric_cols}")
        X = X.drop(columns=non_numeric_cols)

    # Replace inf with NaN
    X.replace([np.inf, -np.inf], np.nan, inplace=True)

    # Drop or fill NaN
    X = X.dropna()
    y = y.loc[X.index]

    clf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
    clf.fit(X, y)

    importances = pd.Series(clf.feature_importances_, index=X.columns)
    importances = importances.sort_values(ascending=False)

    results[attack_name]["classifier"] = clf
    results[attack_name]["importances"] = importances

    print(f"Top 10 features for {attack_name}:")
    print(importances.head(10))
```

```python
# Create RadViz plots for each attack
for attack_name in results.keys():
    print(f"\n{'='*60}")
    print(f"Creating RadViz plots for: {attack_name}")
    print(f"{'='*60}")

    # Get data for this attack
    X = results[attack_name]["X"]
    y = results[attack_name]["y"]
    importances = results[attack_name]["importances"]

    # Select top features for RadViz (RadViz works best with 3-15 features)
    top_features = importances.head(10).index.tolist()

    # Prepare data for RadViz
    X_scaled = prepare_radviz_data(X, y, top_features, attack_name)

    # Sample data if too large (RadViz can be slow with large datasets)
    if len(X_scaled) > 5000:
        # Sample equally from both classes (or as many as available)
        benign_data = X_scaled[X_scaled["Class"] == "Benign"]
        attack_data = X_scaled[X_scaled["Class"] == attack_name]

        # Sample up to 2500 from each class (or all available if less than 2500)
        benign_sample_size = min(2500, len(benign_data))
        attack_sample_size = min(2500, len(attack_data))

        benign_sample = benign_data.sample(n=benign_sample_size, random_state=42)
        attack_sample = attack_data.sample(n=attack_sample_size, random_state=42)
        X_scaled = pd.concat([benign_sample, attack_sample], ignore_index=True)

        print(
            f"Sampled {benign_sample_size} benign and {attack_sample_size} attack samples for visualization"
        )

    # Store prepared data
    results[attack_name]["radviz_data"] = X_scaled
    results[attack_name]["top_features"] = top_features
```

```python
# Generate combined RadViz and feature importance plots
for attack_name in results.keys():
    print(f"Generating combined plot for {attack_name}...")

    X_scaled = results[attack_name]["radviz_data"]
    importances = results[attack_name]["importances"]

    # Create combined plot
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))

    # RadViz plot
    radviz(X_scaled, "Class", ax=ax1, alpha=0.6, s=20)
    ax1.set_title(
        f"RadViz Plot: {attack_name} vs Benign", fontsize=14, fontweight="bold"
    )
    ax1.legend(title="Traffic Type", bbox_to_anchor=(1.05, 1), loc="upper left")
    ax1.grid(True, alpha=0.3)

    # Feature importance plot
    importances.head(10).plot(kind="barh", ax=ax2)
    ax2.set_title(
        f"Top 10 Feature Importances: {attack_name}", fontsize=14, fontweight="bold"
    )
    ax2.set_xlabel("Importance Score")
    ax2.grid(True, alpha=0.3)

    plt.tight_layout()

    # Save the plot
    plot_filename = f"radviz_plots/{attack_name}_radviz_analysis.png"
    plt.savefig(plot_filename, dpi=300, bbox_inches="tight")
    print(f"Combined plot saved as: {plot_filename}")

    # Show plot
    plt.show()
```
