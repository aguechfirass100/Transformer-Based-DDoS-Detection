```python
original_data = "C:/Users/AGFirass/Documents/GitHub/Transformer-Based-DDoS-Detection/data/subset/01-12/flow-level/combined_subset_50k.parquet"

df_original = pd.read_parquet(original_data)
print("#"*10)
print("shape :: ", df_original.shape)
print("#"*10)
print(df_original.columns)
```

```python
df_original.columns = df_original.columns.str.strip()
df_original['Label'].unique()
```

```python
class_names = [
    "DrDoS_DNS",
    "DrDoS_NTP",
    "DrDoS_NetBIOS",
    "TFTP",
    "DrDoS_LDAP",
    "DrDoS_SNMP"
]

df_filtered = df_original[df_original["Label"].isin(class_names)].copy()

df_filtered.reset_index(drop=True, inplace=True)

print("Remaining classes:", df_filtered["Label"].unique())
print("Shape:", df_filtered.shape)
```

```python
pruned_features = [
    "Source IP", "Destination IP", "Timestamp", "ACK Flag Count", "Max Packet Length",
    "Fwd Packets/s", "Destination Port", "Fwd Packet Length Std", "Subflow Fwd Bytes",
    "min_seg_size_forward", "Flow IAT Min", "Protocol", "Packet Length Std",
    "Total Length of Fwd Packets", "Fwd Packet Length Max", "Init_Win_bytes_forward",
    "Fwd IAT Mean", "Fwd IAT Max", "Flow IAT Max", "Flow IAT Mean",
    "Fwd Header Length", "Fwd Packet Length Min", "Average Packet Size",
    "Min Packet Length", "Fwd IAT Total", "Flow Duration"
]

# Keep only pruned features + Label
df_pruned = df_filtered[pruned_features + ["Label"]].copy()

print("Columns kept:", df_pruned.columns.tolist())
print("Shape:", df_pruned.shape)
```

```python
print(df_pruned.dtypes)
```

```python
import pandas as pd
import ipaddress

def ip_to_int(ip: str) -> int:
    """Convert IPv4/IPv6 string to integer."""
    try:
        return int(ipaddress.ip_address(ip))
    except Exception:
        return -1

def process_timestamp_feature(timestamp_value):
    """Convert timestamp to float seconds since epoch."""
    if isinstance(timestamp_value, str):
        try:
            dt = pd.to_datetime(timestamp_value)
            return float(dt.timestamp())
        except:
            return 0.0
    elif isinstance(timestamp_value, pd.Timestamp):
        return float(timestamp_value.timestamp())
    elif isinstance(timestamp_value, (int, float)):
        return float(timestamp_value)
    else:
        return 0.0
```

```python
# Encode IPs
df_pruned["Source IP"] = df_pruned["Source IP"].apply(ip_to_int)
df_pruned["Destination IP"] = df_pruned["Destination IP"].apply(ip_to_int)

# Encode Timestamp
df_pruned["Timestamp"] = df_pruned["Timestamp"].apply(process_timestamp_feature)

# Encode Labels (0–5)
label_mapping = {label: idx for idx, label in enumerate(class_names)}
df_pruned["Label"] = df_pruned["Label"].map(label_mapping)
```

```python
print("Remaining columns and dtypes:")
print(df_pruned.dtypes)
print("\nLabel distribution:")
print(df_pruned["Label"].value_counts())
```

```python
output_path = r"C:\Users\AGFirass\Documents\GitHub\Transformer-Based-DDoS-Detection\notebooks\IFT_data.parquet"
df_pruned.to_parquet(output_path, index=False)
```
