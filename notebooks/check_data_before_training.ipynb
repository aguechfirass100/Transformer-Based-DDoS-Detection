```python
import pandas as pd

# File path
input_file = r"C:\Users\AGFirass\Documents\GitHub\Transformer-Based-DDoS-Detection\notebooks\data_50k.parquet"
output_file = r"C:\Users\AGFirass\Documents\GitHub\Transformer-Based-DDoS-Detection\notebooks\data_20k.parquet"

# Load data
df = pd.read_parquet(input_file)

# Check label distribution
label_col = "Label"
if label_col not in df.columns:
    raise ValueError(f"Column '{label_col}' not found in dataset")

# Target per label
target_per_label = 20000

# Sample data
balanced_df = (
    df.groupby(label_col, group_keys=False)
    .apply(lambda x: x.sample(n=target_per_label, random_state=42))
    .reset_index(drop=True)
)

# Save to new file
balanced_df.to_parquet(output_file, index=False)

output_file
```
