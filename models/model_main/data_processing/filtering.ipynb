```python
import pandas as pd
import pickle

flow_file = r"C:\Users\AGFirass\Documents\GitHub\Transformer-Based-DDoS-Detection\data\01-12\flow-level-parquets\DrDoS_DNS.parquet"
mapping_file = r"/models/model_main/data/mapping/packet_level_mapping/packet_file_flow_mapping_no_duplicates.pkl"

print("Loading flow parquet...")
flow_df = pd.read_parquet(flow_file)
print(f"Original flow-level rows: {len(flow_df)}")

print("Loading flow mapping...")
with open(mapping_file, "rb") as f:
    flow_to_packet_map = pickle.load(f)

valid_flows = set(flow_to_packet_map.keys())
print(f"Unique valid flows from mapping: {len(valid_flows)}")

print("Filtering flows...")
filtered_flow_df = flow_df[flow_df["Flow ID"].isin(valid_flows)].reset_index(drop=True)

print(f"Filtered flow-level rows: {len(filtered_flow_df)}")

output_file = r"C:\Users\AGFirass\Documents\GitHub\Transformer-Based-DDoS-Detection\models\model_main\data\flow_level\DrDoS_DNS_filtered.parquet"
filtered_flow_df.to_parquet(output_file, index=False)

print(f"Saved filtered parquet: {output_file}")
```

# TFTP

```python
import pandas as pd

csv_file = r"C:\Users\AGFirass\Documents\GitHub\Transformer-Based-DDoS-Detection\data\01-12\flow-level\TFTP.csv"
parquet_file = r"/data/01-12/flow-level-parquets/TFTP.parquet"

# Load CSV in chunks to avoid memory issues
chunks = pd.read_csv(csv_file, chunksize=30_000, low_memory=False)

processed_chunks = []
chunk_num = 0

for chunk in chunks:
    chunk_num += 1
    print(f"Processing chunk {chunk_num} with {len(chunk)} rows...")

    # Replace '-' with '_' in Flow ID
    chunk['Flow ID'] = chunk['Flow ID'].str.replace('-', '_')

    # Force SimillarHTTP to string to avoid Parquet type errors
    if 'SimillarHTTP' in chunk.columns:
        chunk['SimillarHTTP'] = chunk['SimillarHTTP'].astype(str)

    processed_chunks.append(chunk)
    print(f"Chunk {chunk_num} processed and appended.")

print("Concatenating all chunks...")
df = pd.concat(processed_chunks)
print(f"Total rows after concatenation: {len(df)}")

print("Saving to Parquet...")
df.to_parquet(parquet_file, index=False)
print("CSV successfully converted to Parquet!")
```

```python
import pyarrow.parquet as pq
import pyarrow as pa
import os
import pickle

flow_file = r"C:\Users\AGFirass\Documents\GitHub\Transformer-Based-DDoS-Detection\data\01-12\flow-level-parquets\TFTP.parquet"
mapping_file = r"/models/model_main/data/mapping/packet_level_mapping/packet_file_flow_mapping_no_duplicates.pkl"
output_file = r"C:\Users\AGFirass\Documents\GitHub\Transformer-Based-DDoS-Detection\models\model_main\data\flow_level\TFTP_filtered.parquet"

# Load the mapping
with open(mapping_file, "rb") as f:
    flow_to_packet_map = pickle.load(f)

valid_flows = set(flow_to_packet_map.keys())
print(f"Number of valid flows in mapping: {len(valid_flows)}\n")

# Open the parquet file
parquet_file = pq.ParquetFile(flow_file)

# Writer will append filtered row groups
writer = None
total_filtered_rows = 0

# Process row groups one by one
for i in range(parquet_file.num_row_groups):
    print(f"--- Processing row group {i+1}/{parquet_file.num_row_groups} ---")

    row_group = parquet_file.read_row_group(i)
    df_group = row_group.to_pandas()

    # print("*"*10)
    # print(df_group.columns)
    # print(df_group["Flow ID"].head())
    # print("*"*10)

    print(f"Original rows in this group: {len(df_group)}")

    # Filter by valid flow IDs
    df_filtered = df_group[df_group["Flow ID"].isin(valid_flows)]
    print(f"Rows kept after filtering: {len(df_filtered)}")

    if len(df_filtered) == 0:
        continue

    # Convert back to Arrow Table
    table_filtered = pa.Table.from_pandas(df_filtered)

    # Initialize writer if first chunk
    if writer is None:
        writer = pq.ParquetWriter(output_file, table_filtered.schema)

    # Write the filtered table
    writer.write_table(table_filtered)

    total_filtered_rows += len(df_filtered)
    print(f"Cumulative filtered rows so far: {total_filtered_rows}\n")

if writer:
    writer.close()

print(f"--Filtered parquet saved: {output_file}")
print(f"Total filtered rows: {total_filtered_rows}")
```
